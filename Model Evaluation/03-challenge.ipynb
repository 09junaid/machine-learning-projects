```python
# Task Name: Regression Model Evaluation Metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Real scores
real_scores = [90,60,80,100]

# model prediction
predicted_score = [85,70,70,95]

mae = mean_absolute_error(real_scores,predicted_score)
mse = mean_squared_error(real_scores,predicted_score)
rmse = np.sqrt(mse)

print(f"MAE: On average off by: {mae}")
print(f"MSE: Squared Mistake Value: {mse}")
print(f"RMSE: Final Realistic Error: {rmse}")
```
