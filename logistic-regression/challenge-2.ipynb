```python
from sklearn.linear_model import LogisticRegression

# Data: X = Spam Keywords Count, Y = 0 (Normal Email) | 1 (Spam Email)
X = [[0], [1], [3], [5], [8]]
Y = [0, 0, 1, 1, 1]

# 1. Create and train the model
model = LogisticRegression()
model.fit(X, Y)

# 2. Get input from the user
keyword_count = float(input("Enter number of spam keywords found in email: "))

# 3. Predict output and calculate probabilities
result = model.predict([[keyword_count]])[0]
probabilities = model.predict_proba([[keyword_count]])[0]

# Calculate percentages
normal_prob = probabilities[0] * 100
spam_prob = probabilities[1] * 100

# 4. Display the final analysis result
print(f"\n--- Analysis Result ---")
if result == 1:
    print(f"Status: SPAM EMAIL ⚠️")
else:
    print(f"Status: NORMAL EMAIL ✅")

print(f"Spam Probability: {spam_prob:.1f}%")
print(f"Normal Probability: {normal_prob:.1f}%")
```
