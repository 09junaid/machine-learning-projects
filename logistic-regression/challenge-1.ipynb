```python
from sklearn.linear_model import LogisticRegression

X = [[1],[2],[3],[4],[5]]  # hours stutied input
Y = [0,0,1,1,1]  # result 0:fail | 1:pass

model =  LogisticRegression()
model.fit(X,Y)

hours = float(input('Enter how many hours you studied: '))
result = model.predict([[hours]])[0]
if result == 1:
    print(f"Based on hour {hours}, you are likely to PASS")
else:
    print(f"Based on hour {hours}, you are likely to FAIL")
```
