```python
# Task Name: Customer Purchase Prediction using KNN Classifier

from sklearn.neighbors import KNeighborsClassifier

# Data: X = Customer Age (2D list), Y = Bought Product (0 = No, 1 = Yes)
X = [[18], [20], [25], [40], [45], [50]]
Y = [0, 0, 0, 1, 1, 1]

# 1. Create KNN model with K=3 (checking 3 nearest neighbors)
model = KNeighborsClassifier(n_neighbors=3)

# 2. Fit/Train the model with data
model.fit(X,Y)

# 3. Get customer age input from user
age = float(input("Enter customer age: "))

# 4. Predict whether the customer will buy the product
prediction = model.predict([[age]])[0]

print(f"\n--- Task: Customer Purchase Prediction ---")
if prediction == 1:
    print(f"Age {age}: Customer WILL buy the product! ✅")
else:
    print(f"Age {age}: Customer WILL NOT buy the product. ❌")
```
