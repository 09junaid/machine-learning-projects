```python
# study hours and attendence %
from sklearn.neighbors import KNeighborsClassifier

# Data: [Study Hours, Attendance %]
X = [
    [1, 30],
    [2, 40],
    [3, 50],
    [6, 80],
    [7, 85],
    [8, 90]
]

# Y: 0 = Fail, 1 = Pass
Y = [0, 0, 0, 1, 1, 1]

# Create model with K=3 (n_neighbors=3)
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, Y)

# Predict for a new student: [4 hours studied, 60% attendance]
new_student = [[4, 60]]
prediction = model.predict(new_student)[0]

if prediction == 1:
    print("Result: PASS 🎉")
else:
    print("Result: FAIL ❌")
```
