```python
# Task Name: Decision Tree Classifier Concept
# For Decision Tree import "DecisionTreeClassifier"

from sklearn.tree import DecisionTreeClassifier

# Data: [Temperature, Humidity]
X = [
    [30, 80],
    [25, 70],
    [35, 90],
    [20, 50],
    [40, 85]
]

# Y : 0 = Don't Play
# Y : 1 = Play Match
Y = [0, 1, 0, 1, 0] 

# 1. Create Model
model = DecisionTreeClassifier()

# 2. Train Model
model.fit(X,Y)

# 3. Predict
new_day=[[22,55]]
prediction = model.predict(new_day)[0]

# 5. Display Final Output
print(f"--- Task: Match Play Decision ---")
print(f"Input Weather -> Temperature: 22°C | Humidity: 55%")

if prediction == 1:
    print("Decision: PLAY MATCH ✅")
else:
    print("Decision: DON'T PLAY 🏠")
```
