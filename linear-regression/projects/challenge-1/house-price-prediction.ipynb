```python
from sklearn.linear_model import LinearRegression
X = [[1000], [1500], [2000], [2500], [3000]]  # sq ft
Y = [50, 75, 95, 125, 145]  # Lakhs
model = LinearRegression()
model.fit(X,Y)
house_size = float(input("Enter house size in sq ft: "))
predicted_price = model.predict([[house_size]])
print(f"\nPredict Price for {house_size} sq ft house:{predicted_price[0]:.2f} Lakhs")
```
