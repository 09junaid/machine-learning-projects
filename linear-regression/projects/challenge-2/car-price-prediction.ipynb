```python
from sklearn.linear_model import LinearRegression

# Data: Car Age in years (X) and Price in Lakhs (Y)
X = [[1], [2], [3], [5], [7]]
Y = [25,22,18,12,6]

model = LinearRegression()
model.fit(X,Y)
car_age = float(input('Enter Car Age (in years): '))
predicted_price = model.predict([[car_age]])
print(f"\nPredict Price for a {car_age} year old car: {predicted_price[0]:.2f} Lakhs")
```
